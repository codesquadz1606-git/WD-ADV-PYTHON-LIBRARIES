Histogram Charts

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(1) # ki jo random numbers create ho rahee h , haar baar unhe same rakhta h.

marks=np.random.normal(65,12,200).clip(0,100) # 200 studnts ka number create kar rahaa h , ensure that number must be b/w 0, 100
# print(marks)

fig,ax=plt.subplots()
ax.hist(marks,bins=[0,40,60,80,100],color="slateblue",edgecolor="white")
ax.set_xlabel("Marks")
ax.set_ylabel("Number Of Students")
plt.show()

fig.savefig("./graphs/histogram.png")
```

```python
batch_a = np.random.normal(60, 10, 200)
batch_b = np.random.normal(72, 8, 200)

fig,ax=plt.subplots()
ax.set_title("Comparing Two Groups",fontweight="bold")
ax.hist(batch_a,edgecolor="white",bins=20,label="Batch A")
ax.hist(batch_b,edgecolor="white",bins=20,label="BAtch B")
ax.legend()

plt.show()

fig.savefig("./graphs/histogram_compare.png")
```

```python
# Mark the Average
fig,ax=plt.subplots()
ax.hist(marks,bins=15,edgecolor="white",color="lightsteelblue")
ax.axvline(marks.mean(),color="red",linestyle="--",label=f"Mean {marks.mean() }")
ax.legend()
plt.show()
fig.savefig("./graphs/histogram_average.png")
```

Scatter Charts

```python
np.random.seed(1)
attendance=np.random.uniform(50,100,60) # 60 students ka data
marks=attendance*0.7+np.random.normal(0,7,60)

fig,ax=plt.subplots()
ax.scatter(attendance,marks)
ax.set_xlabel("Attendance")
ax.set_ylabel("Marks")
plt.show()
```

```python
fees=np.random.uniform(10,50,60)
# print(fees)

fig,ax=plt.subplots()
pts=ax.scatter(attendance,marks,s=fees*5,c=marks,cmap="viridis",alpha=0.7)
fig.colorbar(pts,label="Marks")
plt.show()
```

```python
courses=np.random.choice(["MERN","Data","Python"],60)

fig,ax=plt.subplots()
for name in ["MERN","Data","Python"]:
    pick = courses==name
    ax.scatter(attendance[pick],marks[pick],label=f"{name} ({pick.sum()})")
ax.legend()
plt.show()
```

Pie Charts & Donut Charts

```python
labels=["MERN","Data Analytics","Python"]
students=[45,30,25]

fig,ax=plt.subplots()
ax.pie(students,labels=labels,autopct="%1.0f%%")
plt.show()
```

```python
labels=["MERN","Data Analytics","Python"]
students=[45,30,25]

fig,ax=plt.subplots()
ax.pie(students,labels=labels,autopct="%1.0f%%",explode=[0.08,0,0],colors=["#0E7C6B","#9FD8C9","#D9EFE9"],startangle=90)
plt.show()
```

```python
fig,ax=plt.subplots()
ax.pie(students,labels=labels,autopct="%1.0f%%",wedgeprops=dict(width=0.2))
ax.text(0,0,'100\nstudents',ha="center",va="center",fontsize=15,fontweight="bold")
plt.show()
```
