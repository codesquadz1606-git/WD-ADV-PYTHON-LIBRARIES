Box Plots
A box plot summarises a column in five numbers. It is the fastest way to compare the spread of several groups

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(1)
mern = np.random.normal(70,8,50)
data_analysis = np.random.normal(62,14,50)
python=np.append(np.random.normal(75,5,48),[35,40])

fig,ax=plt.subplots()
ax.boxplot([mern,data_analysis,python],tick_labels=["MERN","DA","Python"])
ax.set_ylabel("Marks")
plt.show()


# MERN : 64.5 (25%) Median :68.5  74.9 (75%)
```

```python
import numpy as np

np.random.seed(1)
mern = np.random.normal(70, 8, 50)
data_analysis = np.random.normal(62, 14, 50)
python = np.append(np.random.normal(75, 5, 48), [35, 40])

for name, marks in [("MERN", mern), ("DA", data_analysis), ("Python", python)]:
    q1, median, q3 = np.percentile(marks, [25, 50, 75])
    iqr = q3 - q1
    low, high = q1 - 1.5 * iqr, q3 + 1.5 * iqr # 64.5-
    outliers = marks[(marks < low) | (marks > high)]
    print(name, q1.round(1), median.round(1), q3.round(1), iqr.round(1), outliers.round(1))
```

IQR : Interquartile Range : Box ki height depend kartii h

```python
# Horizontal Box Plots
fig , ax= plt.subplots()
ax.boxplot([mern,data_analysis,python],tick_labels=["MERN","DA","Python"], orientation="horizontal",patch_artist=True,boxprops=dict(facecolor="lightblue"))

plt.show()
```

```python
# Area CHarts
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
users = [200, 340, 310, 480, 520, 610]

fig,ax=plt.subplots()
ax.plot(months,users,color="teal")
ax.fill_between(months,users,color='teal',alpha=0.2)
plt.show()
```

```python
target = [300, 300, 400, 400, 500, 500]
fig,ax=plt.subplots()
ax.plot(months,users,label="Actual")
ax.plot(months,target,linestyle="--",color="grey",label="Target")
ax.fill_between(months,users,target,color='green',alpha=0.2)
ax.legend()
plt.show()
```

```python
# Stacked Area Chart

mobile = [120, 200, 190, 300, 330, 400]
web = [60, 100, 90, 130, 140, 150]
tablet = [20, 40, 30, 50, 50, 60]

fig,ax=plt.subplots()
ax.stackplot(months,mobile,web,tablet,labels=["Mobile","Web","Tablet"],alpha=0.8)
ax.legend(loc="upper left")
plt.show()
```

```python
cities = ['Delhi', 'Mumbai', 'Pune']
sales = [2130, 1610, 1019]
months = ['Jan', 'Feb', 'Mar', 'Apr']
trend = [120, 150, 135, 180]

fig,(ax1,ax2)=plt.subplots(1,2,figsize=[10,4])
ax1.bar(cities,sales)
ax1.set_title("Sales by City")

ax2.plot(trend,marker='o')
ax2.set_xticks(range(4),months)
ax2.set_title("Monthly Trend")

plt.tight_layout()
plt.show()
```

```python
np.random.seed(0)
fig, ax = plt.subplots(2, 2, figsize=(9, 6))
ax[0, 0].plot(trend, marker='o')
ax[0, 1].bar(cities, sales)
ax[1, 0].hist(np.random.randn(200),edgecolor="white", bins=15)
ax[1, 1].scatter(np.random.rand(30), np.random.rand(30))
fig.suptitle('Business dashboard', fontsize=15, fontweight='bold')
plt.tight_layout()
```

```python
fig,ax=plt.subplots(1,3,figsize=(10,3),sharey=True)
for i , city in enumerate(cities):
    ax[i].bar(["Q1","Q2"],[sales[i]*0.4 , sales[i]*0.6])
    ax[i].set_title(city)
plt.tight_layout()

plt.show()
```
