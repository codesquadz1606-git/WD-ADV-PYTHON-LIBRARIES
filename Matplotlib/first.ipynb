```python
import matplotlib.pyplot as plt

months=["Jul","Aug","Sept"]
revenue=[2398,2987,3912]

plt.plot(revenue,months,marker='o') # creates the data on x-axis & y-axis
plt.title("Monthly Revenue")
plt.xlabel("Months")
plt.ylabel("Revenue")

plt.show()
```

```python
# plt.bar(months,revenue)
plt.barh(months,revenue)
plt.show()
```

```python
plt.pie(revenue,labels=months)
```

```python
plt.scatter(months,revenue)
```

```python
delhi=[1260,630,240]
mumbai=[620,10,990]

plt.plot(months,delhi,marker='o',label='Delhi',color='red')
plt.plot(months,mumbai,marker='o',label='Mumbai',color='green')
plt.legend()

plt.savefig("new.png")
plt.show()
```

```python
import numpy as np
path="./all_images/"
quarters=["Q1","Q2","Q3","Q4"]

# width=0.25
height=0.20

product_A=[20,35,30,40]
product_B=[25,30,35,45]
product_C=[15,25,40,35]

# x=np.arange(len(quarters))
y=np.arange(len(quarters))
# print(x)

# plt.bar(x-width,product_A,0.25,label="product A",color="skyblue")
# plt.bar(x,product_B,0.25,label="product B",color="red")
# plt.bar(x+width,product_C,0.25,label="product C",color="orange")

plt.barh(y-(height/2),product_A,height,label="product A",color="skyblue")
plt.barh(y,product_B,height,label="product B",color="red")
plt.barh(y+(height/2),product_C,height,label="product C",color="orange")


# plt.xlabel("Quarters")
# plt.xticks(x,quarters)
plt.ylabel("Quarters")
plt.yticks(y,quarters)

plt.legend()

plt.savefig(path+"clustered_image.png")

```

```python
path="./all_images/"
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
delhi = [120, 150, 135, 180, 210, 190]
mumbai = [100, 110, 160, 150, 170, 220]
hyderabad=[200,120,450,780,10,90]


fig, ax=plt.subplots()

# fig : this variable shows the figure

ax.plot(months,delhi,linestyle="-",marker='o',label="Delhi")
ax.plot(months,mumbai,linestyle="--",marker='s',label="Mumbai")
ax.plot(months,hyderabad,linestyle=":",marker='*',label="Hyderabad")

ax.set_title("Montly Revenue",fontweight="bold")
ax.set_xlabel("Month")
ax.set_ylabel("Revenue")
ax.legend(loc='upper right')
ax.grid(True,linestyle=":")

fig.savefig(path+"new1.png")

plt.show()
```
