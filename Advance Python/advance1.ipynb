What is Itreators?
An Iterable is any object you can loop over.An Iterator is an object that produces value one at a time and remebers its position.

```python
nums=[10,20,30,40,50]
it=iter(nums)
```

```python
print(it)
```

next() : it is used to iterate the value one by one

```python
next(it)
```

```python
try:
    print(next(it))
except StopIteration as s:
    print(s)
```

Generator: A easy to create the iterator , yeild

```python
def countDown(*n):
    it1=iter(list(n))
    print(it1)

countDown(10,20,30,40,50)
```

```python
def countDown1(start):
    while(start>0):
         yield start #tracks & store every step
        start-=1

```

```python
for n in countDown1(5):
    print(n)
```

```python
import sys

square_list=[x*x for x in range(100000)]

# lazy evaluation & memory optimisation.
square_gen=(x*x for x in range(100000)) # generator expressions

# tuple()


print(type(square_list))
print(type(square_gen))

print(sys.getsizeof(square_list))
print(sys.getsizeof(square_gen))
```

```python
print(square_list[:5])
# print(square_gen[:5])
print(next(square_gen))
```

```python
# Closures : A closures is an inner function that remebers variables from its outer function scope, even after the outer function has finished executing.

def outerFunc():
    msg="Ayush" # outer function scope
    def innerFunc():
        print(msg) # inner function na oy=uter function scope ki value ko access kara.
    return innerFunc()

outerFunc()
```

```python
def multiplier(n):
    def multiply(x):
        return x * n
    return multiply

double=multiplier(5)
double(2)
```
