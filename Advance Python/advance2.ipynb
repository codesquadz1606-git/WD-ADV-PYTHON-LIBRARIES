What is decoraters?

```python
#  How we can pass function inside another function?

def hello(): # I have made an user-defined function
    print("Hello Everyone")

def run(func): # That I can pass any function in parametr & argument.
    func()
    
run(hello)
```

```python
def add(a,b):
    print("Function Started")
    result = a+b
    print("Function Ended")
    return result

def sub(a,b):
    print("Function Started")
    result = a-b
    print("Function Ended")
    return result

def mul(a,b):
    print("Function Started")
    result = a*b
    print("Function Ended")
    return result


# that if we want to change anything in log function, so here I have to change in every function.
```

```python
# This is my decorater function
def logger(func):
    def wrapper():
        print("Function Started")
        func()
        print("Function Ended")
    return wrapper

def say_hello():
    print("Hello Everyone")


hello=logger(say_hello)
hello()
```

```python
# using @ for making a decorater : syntactic sugar
@logger # access the function
def say_hi():
    print("Hii")

say_hi()
```

```python
def logger1(func):
    def wrapper(*args,**kwargs):
        print("Function Started")
        print("Args:",args,"kwargs:",kwargs)
        func(*args,**kwargs)
        print("Function Ended")
    return wrapper
```

```python
@logger1
def add(a,b):
    print(a+b)

add(1,4)
```

```python
@logger1
def sub(a,b):
    print(a-b)

sub(3,1)
sub(a=2,b=3)
```

```python
from functools import wraps
def logger2(func):
    # @wraps(func) # provides the identity to itself.
    def wrapper(*args,**kwargs):
        return func(*args,**kwargs)
    return wrapper

@logger2
def add(a,b):
    """Add two numbers and returns the result"""
    return a+b

print(add.__name__)
print(add.__doc__)
help(add)
```

```python
from functools import wraps

def bold(func):
    @wraps(func)
    def wrapper(*args,**kwargs):
        return "<b>"+func(*args,**kwargs)+"</b>"
    return wrapper

def italic(func):
    @wraps(func)
    def wrapper(*args,**kwargs):
        return "<i>"+func(*args,**kwargs)+"</i>"
    return wrapper

@italic
@bold
def text():
    return "Hello"

print(text())

```
