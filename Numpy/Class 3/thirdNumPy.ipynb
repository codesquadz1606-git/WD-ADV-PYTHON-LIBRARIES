```python
# Flatten , ravel , transpose
m=np.array([[1,2,3,4],[5,6,7,8]])
print(m.flatten())
print(m.ravel()) # this is more faster
print(m.T)
```

```python
# Joining the arrays : concatenate & stack
a=np.array([[1,2],[3,4],[10,11]])
b=np.array([[5,6],[7,8],[12,13]])
c=np.array([[15,16],[17,18],[19,20]])

# new_arr=np.concatenate([a,b],axis=0) # roww add huee
# new_arr=np.concatenate([a,b],axis=1) # columns add huee.

# new_arr = np.vstack([a,b]) # vertical stack (axis =0)
# new_arr = np.hstack([a,b]) # horizaontal stack (axis =1)

new_arr=np.stack([a,b,c]).shape
print(new_arr)
```

```python
# Splitting array
arr1=np.arange(10)
# n1=np.split(arr1,2) # 5 equal parts mae split kardegaa
n1=np.split(arr1,[3,5]) # [0:3],[3:7],[7:]
print(n1)
```
