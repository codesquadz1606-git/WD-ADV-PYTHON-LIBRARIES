Vectorization : It means expressing an operation on whole arrays at once instead of looping over elements in python.

```python
import numpy as np

prices=np.array([100,250,80,400])
qty=np.array([2,1,5,3])

# total = []
# for i in range(len(prices)): # yeh operation slow h.
#     total.append(prices[i]*qty[i])

# Vectorized way
total=prices*qty
gst_total=total+total*0.18
print(gst_total)
```

Universal Functions (ufuncs)


```python
x=np.array([1,4,9,16])
print(np.sqrt(x))
print(np.square(x))
print(np.exp([0,1]))
print(np.log(x)) # log x
print(np.log10(x)) # log 10x
print(np.abs([-3,-2,10]))
print(np.round([2.500,3.891]))

#  two array ufuncs
print(np.add(x,1)) # x+1
print(np.maximum([1,5,6],[4,2,6]))
print(np.minimum([1,5,6],[4,2,6]))
```

```python
# Advance ufuncs 
print(np.add.reduce([1,2,3,4]))
print(np.add.accumulate([1,2,3,4]))
# print()
# print(np.multiply.outer([1,2,3],[1,2]))
print(np.multiply.outer([1,2],[1,2,3]))
```
