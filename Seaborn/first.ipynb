```python
import seaborn as sns
import matplotlib.pyplot as plt

tips=sns.load_dataset('tips')
# data=list(tips['time'])
del tips["size"]
print(type(tips))

# sns.scatterplot(data=tips,x='total_bill',y='tip')
# sns.scatterplot(data=tips,x='day',y='total_bill')
sns.scatterplot(data=tips,x='sex',y='total_bill',hue='day')
# plt.show()
```

```python
sns.relplot(data=tips , x='total_bill',y='tip', col='day')
plt.show()
```

```python
sns.set_theme() # by default : grey background & white grid
tips=sns.load_dataset("tips")
sns.histplot(data=tips,x='total_bill')
plt.show()
```

```python
sns.set_style("ticks")
tips=sns.load_dataset("tips")
sns.histplot(data=tips,x='time')
sns.despine() # removes the top & right
plt.show()
```

```python
sns.set_style("ticks")
```

```python
# Palette :  A set of colrs seaborn use for groups or values.
```

```python
sns.set_theme()

for name in ['deep','pastel','Set2','Blues','rocket','coolwarm']:
    print(name.ljust(7) ," : ", sns.color_palette(name).as_hex())
```

```python
sns.boxplot(data=tips,x='day',y='total_bill',hue='day',palette='coolwarm')
plt.show()
```

```python
sns.set_theme(style='whitegrid')
tips=sns.load_dataset("tips")
# kde(kernel density estimate) 
# sns.histplot(data=tips,x='total_bill',bins=20,kde=True)

# sns.histplot(data=tips,x='total_bill',hue='time',multiple='stack')
sns.histplot(data=tips,x='tip',stat='percent',binwidth=0.5)


plt.show()
```

```python
sns.kdeplot(data=tips,x='total_bill',hue='time',fill=True)

# plt.show()
```

```python
# ECDF : Empirical Cumulative Distribution Function
sns.ecdfplot(data=tips,x='total_bill',hue='size',palette='viridis')
```

```python
sns.countplot(data=tips,x='day',hue='time')
plt.show()
```

```python
sns.barplot(data=tips,x='day',y='total_bill')
print(dir(sns))
plt.show()
```

```python
sns.barplot(data=tips,y='day',x='tip',order=["Sun","Sat","Fri","Thur"],errorbar=None)
plt.show()
```
