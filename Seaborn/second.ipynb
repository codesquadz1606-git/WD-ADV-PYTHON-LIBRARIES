```python
import seaborn as sns
import matplotlib.pyplot as plt

tips=sns.load_dataset('tips')
tips
# print(tips['sex'].value_counts())
```

```python
sns.violinplot(data=tips,x='day',y='total_bill',hue='sex',split=True)
plt.show()

```

```python
# stripplot & swarmplot
fig,ax=plt.subplots(1,2,figsize=(10,3.6))
sns.stripplot(data=tips,x='day',y='total_bill',ax=ax[0])
sns.swarmplot(data=tips,x='day',y='total_bill',ax=ax[1])
ax[0].set_title("stripplot (random jitter)")
ax[1].set_title("swarmplot (no overlapping)")
plt.tight_layout()
plt.show()
```

```python
# Pointplot : means joined by the lines.
# print(tips.groupby("day",)['total_bill'].mean())
print(tips.pivot_table(values='tip',index='day',columns='sex',aggfunc='mean',margins=True).round(2))

males=tips[tips['sex']=="Male"]
print(males.groupby("day")['tip'].max())
print(males.groupby("day")['tip'].min())

sns.pointplot(data=tips,x='day',y='tip',hue='sex',order=['Thur','Fri','Sat','Sun'])
plt.show()
```

```python
# catplot : the figure-level version
sns.catplot(data=tips,x='day',y='total_bill',hue='sex',col='time',kind='box')
plt.show()
```

<!-- Relational Plots -->

```python
# scatterplots
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='time',size='size',style='smoker')
plt.legend(bbox_to_anchor=(1.02,1),loc='upper left')
plt.show()
```

```python
# lineplot
flights=sns.load_dataset("flights")
# flights.head()
# flights.tail()

flights
```

```python
# lineplot
sns.lineplot(data=flights,x='year',y='passengers')
plt.show()
```

```python
sns.lineplot(data=flights,x='year',y='passengers',hue='month',palette='viridis')
plt.legend(ncol=2,fontsize='8')
plt.show()
```

```python
# Regression Plots

sns.regplot(data=tips,x='total_bill',y='tip')
plt.show()
```

```python
# lmplot : regression by groups
sns.lmplot(data=tips,x='total_bill',y='tip',hue='smoker')
plt.show()
```

```python
sns.lmplot(data=tips,x='total_bill',y='tip',col='day')
plt.show()
```

```python
# matrix plot
grid=flights.pivot(index='month',columns='year',values='passengers')
grid.iloc[:3]
```

```python
sns.heatmap(grid,annot=True,annot_kws={'size':7},fmt='.1f',cmap='YlGnBu')

plt.show()
```
