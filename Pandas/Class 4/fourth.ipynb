1. Create a table where missing ages are written as -1 and missing cities as "NA" or "unknown".
2. Replace all of them with real NaN using one replace call with a dictionary.
3. Add a column of prices written like "Rs 1,200" and convert it to a proper integer column.
4. Map a size column of S / M / L to 1 / 2 / 3.

```python
# 1. Create a table where missing ages are written as -1 and missing cities as "NA" or "unknown".
import pandas as pd
import numpy as np

df=pd.DataFrame({
    'name':["Rahul","Priya","Amit","Sneha","Vikram"],
    'age':[25,-1,32,-1,28],
    'city':["Delhi","Na","Noida","unknown","Mumbai"]
})

df
```

```python
# 2. Replace all of them with real NaN using one replace call with a dictionary.
df=df.replace({'age':{-1:np.nan},"city":{"Na":np.nan,'unknown':np.nan}})
df
```

```python
# 3. Add a column of prices written like "Rs 1,200" and convert it to a proper integer column.
df['price'] = ['Rs 1,200', 'Rs 850', 'Rs 12,500', 'Rs 3,000', 'Rs 450']
df["price"]=df['price'].str.replace("Rs","",regex=False).str.replace(",",'',regex=False).str.strip().astype(int)

df["price"]
```

```python
# 4. Map a size column of S / M / L to 1 / 2 / 3.
df['size']=['M','L','S','S','M']
df['size_code']=df['size'].map({'S':1,"M":2,"L":3})
df
```
