```python
import pandas as pd
print(pd.__version__)
```

```python
orders={
    "order_id":[101,102,103,104,105],
    "city":["Delhi","Noida","Gurgaon","Mumbai","Hyderabad"],
    "item":["Iphone 18","Nothing","Redmi","Samsung","Nokia"],
    "amount":[100000,40000,25000,90000,20000]
}


df=pd.DataFrame(orders)
print(df)
```

```python
print(df.shape)
print(df.columns[0])
print(df.head(3))
```

```python
print(df.info())
```

```python
print(df.describe())
```

```python
print(df["amount"])
print(df["city"])

print(df[["order_id","item"]])
```

```python
print("Highest Price :",df["amount"].max())
print("Lowest Price :",df["amount"].min())
print("Total Price:",df["amount"].sum())
print("Average Price :",df["amount"].mean())
```

```python
print(df["amount"]>50000)
```

```python
print(df[df["amount"]>50000])
print(df[df["amount"]<50000])
```

```python
# & , |
print(df[(df["amount"]>50000) & (df["city"]=="Delhi")])
print(df[(df["amount"]>50000) | (df["city"]=="Delhi")])
```
