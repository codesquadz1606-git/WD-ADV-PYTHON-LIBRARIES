Write the full table and the two-column table into two sheets of one Excel file

```python
df=pd.read_csv("./files/students.csv",skiprows=2,na_values=["missing"])
orders=pd.read_csv("./files/orders (1).csv",skiprows=1,na_values={
    'price':[-1],
    'rating':['n.a',-1],
    'parse_dates':['order_date']
})

with pd.ExcelWriter("OverallData.xlsx") as writer:
    df.to_excel(writer,sheet_name="Students",index=False)
    orders.to_excel(writer,sheet_name="Orders",index=False)
```
