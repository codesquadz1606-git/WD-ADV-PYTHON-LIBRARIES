```python
import pandas as pd
from sqlalchemy import create_engine,text
from urllib.parse import quote_plus # this is inbuilt library , parsing the password

pwd=quote_plus("CodeSquadz-1012")
engine=create_engine(f"mysql+pymysql://root:{pwd}@localhost:3306/seconddb")
df=pd.read_sql("SELECT * FROM employees",engine)
df
```

```python
rows=[
    {"id":102,"name":"Sameer","city":"Mumbai","department":"IT","salary":50000},
    {"id":103,"name":"Kashish","city":"Gurgaon","department":"Sales","salary":50000}
]

with engine.begin() as conn:
    conn.execute(text("INSERT INTO employees(id,name,city,department,salary) values(:id,:name,:city,:department,:salary)"),rows)
```

```python
new=pd.DataFrame({"id":[104],"name":["Neha"],"city":["Hyderabad"],"department":["Accounts"],"salary":[35000]})
new.to_sql("employees",engine,if_exists="append",index=False)
```
